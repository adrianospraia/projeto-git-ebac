#### Tabela Verdade

And

```python
print(10 > 2 and 4 < 6)       #V and V - V
print(10 > 2 and 4 > 6)       #V and F - F
print(10 < 2 and 4 < 6)       #F and V - F
print(10 < 2 and 4 > 6)       #F and F - F
```

Or

```python
print(10 > 2 or 4 < 6)        #V or V - V
print(10 > 2 or 4 > 6)        #V or F - V
print(10 < 2 or 4 < 6)        #F or V - V
print(10 < 2 or 4 > 6)        #F or F - F
```

Not

```python
not 10 > 2        #V - F
not 10 < 2        #F - V
```

# Aula 5 - Conheça as estruturas condicionais

Condicional

```python
#If

num = 10

if num == 10:
  print('O valor é igual')
```

```python
#If/Else

num1 = 10
num2 = 11

if num1 > num2:
   (print("o valor 1 é maior que o valor 2"))
else: (print('O valor 1 é menor que o valor 2'))
```

```python
#If / Elif / Else

num1 = 10
num2 = 5

if num1 > num2:
  print('Valor 1 é maior que o valor 2')
elif num1 < num2:
  print('Valor 2 é maior que o valor 1')
else:
  print('Os valores são iguais')
```

Condicionais com operadores lógicos

```python
num1 = 100
num2 = 5
num3 = 20

if num1 > num2 and num1 > num3:
  print('O valor 1 é o maior número da lista')
else:
  print('O valor 1 não é o maior da lista')
```

#Aula 7 - Aplique estruturas de repetição

### For

```python
# For Range
for contador in range(11):
  print(contador)
```

```python
# For Range + Break
for contador in range(1000):
  print(contador)
  if contador == 10:
    break
```

```python
lista = ['Teste 1', 'Teste 2']
```

```python
# For Simples
for var in lista:
  print(var)
```

```python
# for com Enumerate
for ind, var in enumerate(lista):
  print(ind, var)
```

### For com operdores compostos

```python
lista2 = [2, 3, 4]
```

```python
for var in lista2:
  var +=2
  print(var)
```

```python
for var in lista2:
  var*= 2
  print(var)
```

### While

```python
val = 10
```

```python
while val < 20:
  val += 1
  print(val)
```
